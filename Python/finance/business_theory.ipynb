Simplistic simulation.

```python
from collections.abc import Collection
from dataclasses import dataclass
from math import floor

@dataclass
class Business:
    number_of_machines: int = 1
    items_produced_per_machine_per_day: float = 10
    machine_cost: float = 1000.0
    profit_per_item: float = 10.0
    cash: float = 0.0
    total_dividends_given: float = 0.0
    days_since_start: float = 0.0

    def forward(self, days: float = 30.0, fraction_to_reinvest: float = 1.0) -> None:
        assert days >= 0
        new_profit = days * self.number_of_machines * self.items_produced_per_machine_per_day * self.profit_per_item
        to_reinvest = fraction_to_reinvest * new_profit
        print(f"Earned {new_profit}, reinvesting {to_reinvest}, distributing {new_profit - to_reinvest}.")
        self.cash += to_reinvest
        self.total_dividends_given += new_profit - to_reinvest
        if self.cash >= self.machine_cost:
            new_machines, remaining_cash = floor(self.cash // self.machine_cost), self.cash % self.machine_cost
            print(f"Buying {new_machines} new machine(s).")
            self.cash = remaining_cash
            self.number_of_machines += new_machines
        self.days_since_start += days

    def status(self) -> None:
        print(f"days_since_start={self.days_since_start}\ttotal_dividends_given={self.total_dividends_given}\tcash={self.cash}\tnumber_of_machines={self.number_of_machines}")

def simulate(n_steps: int, fraction_to_reinvest: float = 1.0):
    b = Business()
    b.status()
    for k in range(n_steps):
        b.forward(fraction_to_reinvest=fraction_to_reinvest)
        b.status()

def multiple_simulations(fractions_to_reinvest: Collection[float] = ()):
    for fraction_to_reinvest in fractions_to_reinvest:
        print(f"\n{fraction_to_reinvest=}")
        simulate(10, fraction_to_reinvest)

multiple_simulations((.0, .5, 1.))
```
